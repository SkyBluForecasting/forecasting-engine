```python
import pandas as pd
import numpy as np
#os is used for MacOS paths
import os
#The glob module is used to find all file names matching a pattern,
from glob import glob

import geopandas as gpd
```

```python
folder_path = '/Users/will/Documents/Beacon MVP/Projects/IESO Data Set Up/Weather stations to FSAs'

# --- Step 1: Load the FSA shapefile ---
path_to_shapefile = (os.path.join(folder_path, 'lfsa000a21a_e', 'lfsa000a21a_e.shp'))
fsa_gdf = gpd.read_file(path_to_shapefile)

print(fsa_gdf.crs)  # Should output EPSG:3347

# Reproject to EPSG:4326 (WGS 84 lat/lon)
fsa_gdf_4326 = fsa_gdf.to_crs(epsg=4326)

# Save to a new shapefile (optional)
fsa_gdf_4326.to_file("/Users/will/Documents/Beacon MVP/Projects/IESO Data Set Up/Weather stations to FSAs/Converted CRS Shp file/fsa_4326.shp")
```
